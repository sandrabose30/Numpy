2D Arrays – Business Reports (21️–30)
Use the matrix:
[[100,120,140],
[150,160,170],
[200,210,220]]

```python
matrix = np.array([[100, 120, 140],
                   [150, 160, 170],
                   [200, 210, 220]])
print(matrix)
# 21. Print the second store sales.
print("2nd Store Sales:", matrix[1])

print("\n")

# 22. Print sales of Product C across stores.
print("Product C Sales:", matrix[:, 2])

print("\n")

# 23. Print the sales value at row 2 column 3.
print("Sales at Row 2, Column 3:", matrix[1, 2])

print("\n")

# 24. Replace the center sales value with 999.
matrix_copy = matrix.copy()
matrix_copy[1, 1] = 999
print(matrix_copy)

print("\n")

# 25. Extract sales data of first two stores.
print(matrix[:2, :])

print("\n")

# 26. Extract sales of the last product.
print(matrix[:, -1])

print("\n")

# 27. Find all sales values greater than 1️60.
print(matrix[matrix > 160])

print("\n")

# 28. Calculate total sales per store (row sum).
print("Total sales per store:", np.sum(matrix, axis=1))

print("\n")

# 29. Calculate total sales per product (column sum).
print("Total sales per product:", np.sum(matrix, axis=0))

print("\n")

# 30. Find the maximum sales value per store.
print("Max sales per store:", np.max(matrix, axis=1))
```

```python
# ⃣Reshaping Arrays – Analytics Data (31️–35)

# 31. Create an array representing 1️2 months revenue and reshape into a 3×4 quarterly matrix.
monthrevenue = np.arange(1, 13) * 10000
matrix = monthrevenue.reshape(3, 4)
print(matrix)

print("\n")

# 32. Convert the matrix
# [[1000,2000,3000],
# [4000,5000,6000]]
# into a 1️D revenue array.
matrix = np.array([[1000, 2000, 3000],
                   [4000, 5000, 6000]])
flat_revenue = matrix.flatten()
print(flat_revenue)

print("\n")

# 33. Create an array representing 1️6 hourly server requests and reshape into a 4×4 matrix. ⃣
requests = np.arange(1, 17)
matrix = requests.reshape(4, 4)
print(matrix)

print("\n")

# 34. Convert a 3×4 sales matrix into a 4×3 matrix.
sales = np.arange(1, 13).reshape(3, 4)
print("Original 3x4 Matrix:\n", sales)
sales = sales.reshape(4, 3)
print("\nReshaped 4x3 Matrix:\n", sales)

print("\n")

# 35. Flatten the matrix representing product sales
# [[120,150],
# [200,250],
# [300,350]]

matrix = np.array([[120, 150],
                         [200, 250],
                         [300, 350]])
sales = matrix.flatten()
print("Flattened Sales:", sales)

print("\n")

# 36. Add daily orders from two platforms
# [100,200,300]
# [50,150,250]
platform1 = np.array([100, 200, 300])
platform2 = np.array([50, 150, 250])
total_orders = platform1 + platform2
print(total_orders)

print("\n")

# 37. Increase product prices by multiplying with tax factor 1️.1️
# [1000,2000,3000]
prices = np.array([1000, 2000, 3000])
taxed = prices * 1.1
print(taxed)

print("\n")

# 38. Find the square of customer purchase quantities
# [2,3,4,5]
quantities = np.array([2, 3, 4, 5])
squared = quantities ** 2
print(squared)

# 39. Calculate profit by subtracting cost from revenue
# revenue = [5000,6000,7000]
# cost = [3000,3500,4000]
revenue = np.array([5000, 6000, 7000])
cost = np.array([3000, 3500, 4000])
profit = revenue - cost
print(profit)

print("\n")

# 40. Calculate discounted prices by dividing by 2
# [2000,4000,6000]
prices = np.array([2000, 4000, 6000])
discounted = prices / 2
print(discounted)

print("\n")

# 41. Find the total daily revenue 
# [12000,15000,18000,20000]
revenue = np.array([12000, 15000, 18000, 20000])
total_revenue = np.sum(revenue)
print("Total Revenue:", total_revenue)

print("\n")

# 42. Find the maximum and minimum product prices
# [450,1200,700,300,900]
prices = np.array([450, 1200, 700, 300, 900])
print("Max Price:", np.max(prices))
print("Min Price:", np.min(prices))

print("\n")
```

```python
# 6⃣Statistical Operations – Data Analytics (43–47)

# 43. Find the average daily sales 
# [12000,15000,18000,20000,17000]

daily_sales = np.array([12000, 15000, 18000, 20000, 17000])
avg_sales = np.mean(daily_sales)
print("Average Daily Sales:", avg_sales)

print("\n")

# 44. Find the median customer purchase value
# [500,700,900,1200,1500]
purchases = np.array([500, 700, 900, 1200, 1500])
median = np.median(purchases)
print("Median Purchase Value:", median)

print("\n")

# 45. Calculate variance of monthly revenue
# [10000,12000,14000,16000]
monthly = np.array([10000, 12000, 14000, 16000])
revenue_variance = np.var(monthly)
print("Revenue Variance:", revenue_variance)

print("\n")

# 46. Calculate standard deviation of product ratings
# [3,4,5,4,5]
ratings = np.array([3, 4, 5, 4, 5])
ratings_std = np.std(ratings)
print("Standard Deviation:", ratings_std)

print("\n")

# 47. Find the total revenue from a 3×3 sales matrix.
sales_matrix = np.array([[100, 120, 140],
                         [150, 160, 170],
                         [200, 210, 220]])

total_revenue = np.sum(sales_matrix)
print("Total Revenue:", total_revenue)
```

```python
# Searching & Unique Values – Analytics (48–50)
# 48. Find the index of the highest product price
# [1200,2500,3000,800,1500]
prices = np.array([1200, 2500, 3000, 800, 1500])
index = np.argmax(prices)
print("Index of highest price:", index)

print("\n")

# 49. Find the index of lowest order value
# [500,1200,750,2000,1500]
orders = np.array([500, 1200, 750, 2000, 1500])
min_index = np.argmin(orders)
print("Index of lowest order value:", min_index)

print("\n")

# 50. Find unique product categories
# ["electronics","fashion","electronics","grocery","fashion"]
categories = np.array(["electronics", "fashion", "electronics", "grocery", "fashion"])
unique = np.unique(categories)
print("Unique categories:", unique)
```
