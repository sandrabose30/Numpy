1️⃣ Why NumPy? (Speed vs Python Lists)
Python lists:

Store elements in different memory locations
Can store mixed data types
Each element is a Python object
Operations run through Python interpreter (slow for large data)
NumPy:

Stores data in contiguous memory
Same datatype
Implemented in C
Supports vectorized operations (very fast)

```python
import numpy as np
```

```python
# Python list addition (slow approach)
a = [1,2,3]
b = [4,5,6]
result = []
for i in range(len(a)):
    result.append(a[i] + b[i])
print(result)
```

```python
# NumPy vectorized addition (fast approach)
import numpy as np
a = np.array([1,2,3])
b = np.array([4,5,6])
print(a + b)
```

```python
# Large scale computation example
a = np.arange(1000000)
b = np.arange(1000000)
c = a + b
print('Computation completed')
```

##3️⃣ Creating Arrays
np.array() → Converts a Python list or tuple into a NumPy array for fast computations.

np.zeros() → Creates a NumPy array filled entirely with zeros.

np.ones() → Creates a NumPy array filled entirely with ones.

np.arange() → Generates a NumPy array with evenly spaced values within a specified range (step-defined).

np.linspace() → Generates a NumPy array with a specified number of evenly spaced values between two endpoints.

```python
print(np.zeros((2,3)))
print(np.ones((3,2)))
print(np.arange(0,10,2))
print(np.linspace(0,1,5))
print(np.eye(3,4))
```

```python
x = np.arange(6)
print(x)
```

4️⃣ Array Attributes¶
ndim → Returns the number of dimensions (axes) of the array.

shape → Returns the dimensions of the array as a tuple (rows, columns, …).

size → Returns the total number of elements in the array.

dtype → Returns the data type of the array elements.

itemsize → Returns the memory size (in bytes) of each array element.

```python
a = np.array([[1,2,3],[4,5,6]])
print(a)
print(a.ndim)
print(a.shape)
print(a.size)
print(a.dtype)
print(a.itemsize)
```
